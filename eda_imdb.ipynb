```python
import pandas as pd

DATA_PATH = "backend/data/raw/imdb_movies.json"
```

```python
# detta funkar inte, det har visat sig att detta är inte riktigt json-format
# datat ser ut såhär {'id': 'tt0959285', 'primaryTitle': 'Azhagiya Tamilmagan', ...}
# json skall vara {"id": "tt0959285", "primaryTitle": "Azhagiya Tamilmagan",
# inte enkla ' utan "
# men datan är formaterad som python dictionaries

movies_df = pd.read_json(DATA_PATH, lines=True)
movies_df.head(5)
```

```python
import ast  # denna modulen behövs för att kunna använda funktionen ast.literal_eval()
            # den kan läsa text som ser ut som vanliga python-datastrukturer
```

Jag måste bygga en dataframe av data

```python
# en film per rad, och varje rad är formaterad som en Python-dict

# väg framåt: Fil --> läs en rad --> konvertera till dict --> samla alla dicts --> skapa en dataframe

```

```python
# Tom lista där varje film skall sparas
movies = []

# öppnat filen med i read-läge med UTF-8 encoding
with open(DATA_PATH, "r", encoding="utf-8") as file:
    
    #läser filen rad för rad
    for line in file:
        
        # konverterar varje rad från en sträng till en pyton-dict
        movie = ast.literal_eval(line)
        
        # lägger till filmen till lista
        movies.append(movie)
        
# skapar en dataframe från listan med dictionaries
movies_df = pd.DataFrame(movies)

    
```
